```python
from pgmpy import inference
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination

model=DiscreteBayesianNetwork([
    ('Rain','Sprinkler'),
    ('Rain','GrassWet'),
    ('Sprinkler','GrassWet')
])

#P(Rain)
cpd_rain=TabularCPD(
    variable='Rain', variable_card=2, values=[[0.8],[0.2]]
)
#P(Sprinkler| rain)
cpd_sprinkler=TabularCPD(
    variable='Sprinkler',variable_card=2, values=[[0.6,0.99],[0.4,0.01]],
    evidence=['Rain'], evidence_card=[2]
)
#P(GrassWet| Sprinkler,Rain)
cpd_grass=TabularCPD(
    variable='GrassWet', variable_card=2, values=[[1.0,0.2,0.1,0.01],[0.0,0.8,0.9,0.99]],
    evidence=['Sprinkler','Rain'], evidence_card=[2,2]
)

model.add_cpds(cpd_rain,cpd_sprinkler,cpd_grass)
print("Model is valid: ", model.check_model())

inference=VariableElimination(model)

result = inference.query(
    variables=['Rain'], evidence={'GrassWet': 1}
)
print(result)
```
